```python
# ensure annotations

def get_product(x: int, y: int) -> int:
    return x * y
```

```python
get_product(2, "33")
```

```python
from ensure import ensure_annotations

@ensure_annotations
def get_product(x: int, y: int) -> int:
    return x * y

get_product(2, 3)
```
